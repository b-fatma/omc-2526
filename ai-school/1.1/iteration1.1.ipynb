## 04. Operators

### Arithmetic Operators

````python
a = 10
b = 3

print(a + b)   # addition
print(a - b)   # subtraction
print(a * b)   # multiplication
print(a / b)   # division
print(a // b)  # floor division
print(a % b)   # modulo
print(a ** b)  # exponentiation
````

### Comparison Operators

````python
x = 5
y = 10

print(x == y)   # equal to
print(x != y)   # not equal to
print(x > y)    # greater than
print(x < y)    # less than
print(x >= y)   # greater than or equal
print(x <= y)   # less than or equal
````

### Logical Operators

````python
a = True
b = False

print(a and b)  # both must be True
print(a or b)   # at least one is True
print(not a)    # negation
````

````python
# Practical example
age = 20
is_student = True

if age >= 18 and is_student:
    print("Adult student")
````

## 05. Control Flow

### If Statements

````python
age = 18

if age >= 18:
    print("Adult")
else:
    print("Minor")
````

### If-Elif-Else

````python
score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
elif score >= 60:
    grade = "D"
else:
    grade = "F"

print(f"Grade: {grade}")
````

### Nested If Statements

````python
age = 20
has_license = True

if age >= 18:
    if has_license:
        print("Can drive")
    else:
        print("Need license")
else:
    print("Too young to drive")
````

### Ternary Operator (One-line If-Else)

````python
age = 20
status = "Adult" if age >= 18 else "Minor"
status
````

### For Loop

````python
# Loop through range
for i in range(5):
    print(i)
````

````python
# Range with start and end
for i in range(2, 7):
    print(i)
````

````python
# Range with step
for i in range(0, 10, 2):
    print(i)
````

### While Loop

````python
count = 0
while count < 5:
    print(count)
    count += 1 # Important! Avoid infinite loops
````

### Break and Continue

````python
# Break - exit loop
for i in range(10):
    if i == 5:
        break
    print(i)  
````

````python
# Continue - skip iteration
for i in range(10):
    if i % 2 == 0:
        continue
    print(i) 
````

### Exception Handling

````python
# Try-except
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)
except ValueError:
    print("That's not a valid number!")
except ZeroDivisionError:
    print("Can't divide by zero!")
except Exception as e:
    print(f"An error occurred: {e}")
finally:
    print("This always executes")
````

### Lambda Functions

````python
# Regular function
def square(x):
    return x ** 2

# Lambda equivalent
square = lambda x: x ** 2

print(square(5))  # 25

# Common use: with map, filter
numbers = [1, 2, 3, 4, 5]

# Map - apply function to all elements
squared = list(map(lambda x: x**2, numbers))
print(squared)

# Filter - keep elements that match condition
evens = list(filter(lambda x: x % 2 == 0, numbers))
print(evens)
````
