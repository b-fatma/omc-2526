## 02. Series: The Building Block

### a. Creating Series

````python
# From list
s = pd.Series([10, 20, 30, 40])
s
````

````python
# From dictionary
s = pd.Series({'a': 10, 'b': 20, 'c': 30})
s
````

````python
# With custom index
s = pd.Series([10, 20, 30], index=['x', 'y', 'z'])
s
````

````python
# From NumPy array
s = pd.Series(np.array([1, 2, 3, 4, 5]))
s
````

### b. Series Attributes

- `s.values` - underlying NumPy array
- `s.index` - index labels
- `s.dtype` - data type
- `s.shape` - dimensions
- `s.size` - number of elements

````python
s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
````

````python
print("Series:")
print(s)
print()

print(f"s.values → {s.values}")
print(f"s.index  → {s.index.tolist()}")
print(f"s.dtype  → {s.dtype}")
print(f"s.shape  → {s.shape}")
print(f"s.size   → {s.size}")
````

### c. Series Operations

- Indexing: `s[0]`, `s['label']`
- Slicing: `s[1:3]`, `s['a':'c']`
- Boolean indexing: `s[s > 20]`
- Basic statistics: `s.mean()`, `s.sum()`, `s.max()`

````python
print("Series:")
print(s)
print()

print(f"s.iloc[0]        → {s.iloc[0]}")
print(f"s['c']      → {s['c']}")
print()

print(f"s.iloc[1:3]      → {s.iloc[1:3].tolist()}")
print(f"s['a':'c']  → {s['a':'c'].tolist()}")
print()

print(f"s[s > 20]   → {s[s > 20].tolist()}")
print()

print(f"s.mean()    → {s.mean()}")
print(f"s.sum()     → {s.sum()}")
print(f"s.max()     → {s.max()}")
````

## 05. Indexing and Selection

### a. Selecting Columns

````python
sales_df.head()
````

````python
# Single column (returns Series)
sales_df['product']
````

````python
# Multiple columns (returns DataFrame)
sales_df[['product', 'price']]
````

### b. Selecting Rows by Position: `.iloc[]`

````python
# Single row
sales_df.iloc[0]
````

````python
# Multiple rows
sales_df.iloc[0:3]         # slicing
sales_df.iloc[[0, 2, 4]]   # multiple indices
````

````python
# Rows and columns
sales_df.iloc[0:3, 1:3]    
sales_df.iloc[:, 0] 
````

### c. Selecting Rows by Label: `.loc[]`

````python
# By index label
sales_df.loc[1001]            # row with index label order_id=1001
sales_df.loc[1001:1004]          # inclusive slicing!
````

````python
# By condition
sales_df.loc[sales_df['quantity'] > 3]
````

````python
# Rows and columns by label
sales_df.loc[1001:1005, 'product':'city']
# sales_df.loc[:, 'city'] 
````

### d. Boolean Indexing (Filtering)

````python
# Single condition
sales_df[sales_df['price'] > 25]
````

````python
# Multiple conditions (use ~, & and | with parentheses!)
sales_df[(sales_df['quantity'] > 1) & (sales_df['price'] < 100)]
sales_df[(sales_df['quantity'] == 1) | (sales_df['price'] > 500)]
````

````python
# Using .isin()
sales_df[sales_df['city'].isin(['New York', 'Los Angeles', 'Chicago'])]
````

````python
# Using .str methods
sales_df[sales_df['customer_name'].str.startswith('B')]
sales_df[sales_df['customer_name'].str.contains('son', na=False)]
sales_df[sales_df['product'].str.contains('top', case=False, na=False)]
````

## 06. Data Cleaning Essentials

### a. Handling Missing Data

**Detecting Missing Values:**

````python
sales_df.isnull()          # Boolean DataFrame
````

````python
sales_df.isnull().sum()    # Count per column
````

````python
sales_df.isnull().any()    # Boolean per column
````

````python
# Alternative
sales_df.isna()            # same as isnull()
````

**Removing Missing Values:**

````python
# Drop rows with any missing values
df_clean = sales_df.dropna()
````

````python
# Drop rows where specific column is missing
df_clean = sales_df.dropna(subset=['city'])
````

````python
# Drop columns with any missing values
df_clean = sales_df.dropna(axis=1)
````

````python
# Drop rows only if ALL values are missing
df_clean = sales_df.dropna(how='all')

# Drop rows if more than n non-NA values are missing
df_clean = sales_df.dropna(thresh=3)
````

**Filling Missing Values:**

````python
# Fill with a specific value
sales_df['price'].fillna(0)
sales_df.fillna(0)  # all columns

# Fill with column mean/median
sales_df['price'].fillna(sales_df['price'].mean())
sales_df['price'].fillna(sales_df['price'].median())

# Forward fill (use previous value)
sales_df['price'].ffill()

# Backward fill (use next value)
sales_df['price'].bfill()

# Fill different columns differently
sales_df.fillna({'price': 0, 'city': 'Unknown', 'payment_method': 'Not Specified'})
````

### b. Removing Duplicates

````python
# Check for duplicates
sales_df.duplicated()           # Boolean Series
sales_df.duplicated().sum()     # Count
````

````python
# View duplicate rows
sales_df[sales_df.duplicated()]
````

````python
# Keep last occurrence instead of first
df_clean = sales_df.drop_duplicates()
df_clean
````

````python
# Based on specific columns
df_clean = sales_df.drop_duplicates(subset=['product', 'customer_name'])
````

### c. Data Type Conversions

````python
data = {
    'id': [1001, 1002, 1003, 1004, 1005],
    'age': [25.0, 30.0, 35.0, 28.0, 32.0],  # stored as float, should be int
    'score': ['85.5', '90.0', '78.3', '92.1', '88.7'],  # stored as string, should be float
    'price': ['99.99', '149.50', '79.00', '120.25', '199.99'],  # stored as string, should be float
    'date': ['2024-01-15', '2024-01-16', '2024-01-17', '2024-01-18', '2024-01-19']  # stored as string, should be datetime
}

df = pd.DataFrame(data)
````

````python
df.dtypes
````

````python
df['age'] = df['age'].astype(int)
df['price'] = df['price'].astype(float)
df['date'] = pd.to_datetime(df['date'])
df = df.astype({'score': float, 'id': str})
````

````python
df.dtypes
````
